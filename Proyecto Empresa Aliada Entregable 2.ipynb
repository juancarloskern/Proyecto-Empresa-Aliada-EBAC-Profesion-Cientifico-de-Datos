```python
# Paso 1: Cargar el archivo Consolidated_Data.csv en Pandas

# Importamos la librería necesaria
import pandas as pd

# Cargamos el archivo CSV en un DataFrame
df_consolidado = pd.read_csv('Consolidated_Data.csv')

# Mostramos las primeras filas para verificar que se cargó correctamente
df_consolidado.head()
```

```python
# Paso 2: Identificar las variables clave para el análisis

# Revisamos la estructura y tipos de datos del DataFrame
df_consolidado.info()
# Verificamos si hay valores nulos en las columnas
df_consolidado.isnull().sum()
# Obtenemos estadísticas descriptivas de las variables numéricas
df_consolidado.describe()
# Revisamos el número de registros por región para entender el reparto geográfico
df_consolidado['region'].value_counts()
# Revisamos el número de registros por fabricante para identificar los principales jugadores
df_consolidado['manufacturer'].value_counts()
# Revisamos el número de registros por marca para identificar el portafolio de productos
df_consolidado['brand'].value_counts()
# Revisamos la correlación entre las variables numéricas para entender relaciones entre ventas en unidades y en valor
df_consolidado[['total_unit_sales', 'total_value_sales', 'total_unit_avg_weekly_sales']].corr()
```

```python
# Paso 3: Visualizar la distribución de ventas

# Importamos las librerías necesarias para los gráficos
import matplotlib.pyplot as plt
import seaborn as sns

# Histograma de la distribución de ventas en valor
plt.figure(figsize=(10,6))
sns.histplot(df_consolidado['total_value_sales'], bins=100, kde=True)
plt.title('Distribución de Ventas en Valor')
plt.xlabel('Ventas en Valor')
plt.ylabel('Frecuencia')
plt.savefig('3.1-Distribución de Ventas en Valor.png', dpi=300, bbox_inches='tight')
plt.show()

# Histograma de la distribución de ventas en unidades
plt.figure(figsize=(10,6))
sns.histplot(df_consolidado['total_unit_sales'], bins=100, kde=True)
plt.title('Distribución de Ventas en Unidades')
plt.xlabel('Ventas en Unidades')
plt.ylabel('Frecuencia')
plt.savefig('3.2-Distribución de Ventas en Unidades.png', dpi=300, bbox_inches='tight')
plt.show()

# Boxplot de ventas en valor segmentado por marca
plt.figure(figsize=(12,6))
sns.boxplot(x='brand', y='total_value_sales', data=df_consolidado)
plt.title('Distribución de Ventas en Valor por Marca')
plt.xlabel('Marca')
plt.ylabel('Ventas en Valor')
plt.xticks(rotation=45)
plt.savefig('3.3-Distribución de Ventas en Valor por Marca.png', dpi=300, bbox_inches='tight')
plt.show()

# Boxplot de ventas en unidades segmentado por región
plt.figure(figsize=(12,6))
sns.boxplot(x='region', y='total_unit_sales', data=df_consolidado)
plt.title('Distribución de Ventas en Unidades por Región')
plt.xlabel('Región')
plt.ylabel('Ventas en Unidades')
plt.xticks(rotation=45)
plt.savefig('3.4-Distribución de Ventas en Unidades por Región.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Distribución de Ventas en Valor
La mayoría de los productos generan ventas en valor bajas.
Hay outliers que indican productos estrella con ventas significativamente altas.

### Distribución de Ventas en Unidades
La venta en unidades se concentra en valores bajos, indicando una oferta de productos de baja rotación o nichos específicos.
Existen picos de ventas, probablemente debidos a campañas o promociones.

### Ventas en Valor por Marca
CLORALEX y CLOROX lideran con ventas en valor altas y dispersas.
Marcas como VANISH, SHOUT y otras tienen un comportamiento de ventas más uniforme, pero bajo volumen de ventas en valor.

### Ventas en Unidades por Región
TOTAL AUTOS SCANNING MEXICO muestra el mayor volumen de ventas en unidades.
Las otras regiones tienen menor participación y comportamientos consistentes.

### Distribución de Ventas en Unidades
Se comporta de manera similar a total_value_sales, aunque los outliers son

### Ventas en Valor por Marca
Las marcas CLORALEX, CLOROX y BLANCATEL muestran ventas más elevadas, pero también una variabilidad muy alta (outliers significativos).

### Ventas en Unidades por Región

### Violinplots
Complementan bien los boxplots, mostrando la distribución bimodal o multimodal de ciertas marcas y regiones (por ejemplo, CLORALEX y BLANCATEL tienen distribuciones más extendidas).

```python
# Paso 4: Analizar la tendencia de ventas en el tiempo

import matplotlib.pyplot as plt
import seaborn as sns

# Aseguramos que la fecha esté en datetime
df_consolidado['date'] = pd.to_datetime(df_consolidado['date'])
df_consolidado.sort_values(by='date', inplace=True)

# Agrupación por fecha
ventas_por_fecha = df_consolidado.groupby('date').agg({
    'total_value_sales': 'sum',
    'total_unit_sales': 'sum'
}).reset_index()

# Gráfico general ventas en valor
plt.figure(figsize=(14, 6))
sns.lineplot(data=ventas_por_fecha, x='date', y='total_value_sales')
plt.title('Tendencia de Ventas en Valor a lo largo del tiempo')
plt.xlabel('Fecha')
plt.ylabel('Ventas en Valor')
plt.xticks(rotation=45)
plt.savefig('4.1-Tendencia de Ventas en Valor a lo largo del tiempo.png', dpi=300, bbox_inches='tight')
plt.show()

# Gráfico general ventas en unidades
plt.figure(figsize=(14, 6))
sns.lineplot(data=ventas_por_fecha, x='date', y='total_unit_sales')
plt.title('Tendencia de Ventas en Unidades a lo largo del tiempo')
plt.xlabel('Fecha')
plt.ylabel('Ventas en Unidades')
plt.xticks(rotation=45)
plt.savefig('4.2-Tendencia de Ventas en Unidades a lo largo del tiempo.png', dpi=300, bbox_inches='tight')
plt.show()

# Tendencia por región
ventas_por_fecha_region = df_consolidado.groupby(['date', 'region']).agg({
    'total_value_sales': 'sum'
}).reset_index()

plt.figure(figsize=(14, 6))
sns.lineplot(data=ventas_por_fecha_region, x='date', y='total_value_sales', hue='region')
plt.title('Tendencia de Ventas en Valor por Región a lo largo del tiempo')
plt.xlabel('Fecha')
plt.ylabel('Ventas en Valor')
plt.legend(title='Región')
plt.xticks(rotation=45)
plt.savefig('4.3-Tendencia de Ventas en Valor por Región a lo largo del tiempo.png', dpi=300, bbox_inches='tight')
plt.show()

# Tendencia por marca
ventas_por_fecha_marca = df_consolidado.groupby(['date', 'brand']).agg({
    'total_value_sales': 'sum'
}).reset_index()

marcas_principales = ventas_por_fecha_marca['brand'].value_counts().head(5).index.tolist()
ventas_marcas_filtradas = ventas_por_fecha_marca[ventas_por_fecha_marca['brand'].isin(marcas_principales)]

plt.figure(figsize=(14, 6))
sns.lineplot(data=ventas_marcas_filtradas, x='date', y='total_value_sales', hue='brand')
plt.title('Tendencia de Ventas en Valor por Marca a lo largo del tiempo (Top 5 Marcas)')
plt.xlabel('Fecha')
plt.ylabel('Ventas en Valor')
plt.legend(title='Marca')
plt.xticks(rotation=45)
plt.savefig('4.4-Tendencia de Ventas en Valor por Marca a lo largo del tiempo.png', dpi=300, bbox_inches='tight')
plt.show()

# Promedio móvil opcional
ventas_por_fecha['rolling_value_sales'] = ventas_por_fecha['total_value_sales'].rolling(window=4).mean()

plt.figure(figsize=(14, 6))
sns.lineplot(data=ventas_por_fecha, x='date', y='rolling_value_sales')
plt.title('Tendencia de Ventas en Valor con Promedio Móvil (4 semanas)')
plt.xlabel('Fecha')
plt.ylabel('Promedio Móvil de Ventas en Valor')
plt.savefig('4.5-Tendencia de Ventas en Valor con Promedio Móvil.png', dpi=300, bbox_inches='tight')
plt.xticks(rotation=45)
plt.show()
```

### Tendencia general de ventas en valor y unidades
Se observa una variabilidad alta, con picos recurrentes en ventas de valor y unidades.
Hay temporadas o eventos especiales donde las ventas se disparan (ej. mediados de 2022 y 2023).

### Por Región
La región "TOTAL AUTOS SCANNING MÉXICO" concentra la mayoría de las ventas, mientras que otras áreas tienen ventas estables y mucho menores.
Esto podría indicar una mayor cobertura comercial o mejor penetración de mercado en esa región.

### Por Marca
VANISH es la marca líder en ventas de valor a lo largo del tiempo.
Otras marcas como BANDERA, SHOUT, BCOLORHADA tienen ventas mucho menores en comparación.

### Promedio Móvil
El uso de un promedio móvil de 4 semanas permite ver tendencias estacionales que no son claras en la serie sin suavizar.
Se observa una recuperación en el valor de ventas a partir del segundo trimestre de 2023.

```python
# Paso 5: Explorar la relación entre diferentes variables.

import matplotlib.pyplot as plt
import seaborn as sns

# Scatter Plot 1: Ventas en unidades vs. Ventas en valor
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_consolidado, x='total_unit_sales', y='total_value_sales', alpha=0.5)
plt.title('Relación entre Ventas en Unidades y Ventas en Valor')
plt.xlabel('Ventas en Unidades')
plt.ylabel('Ventas en Valor')
plt.savefig('5.1-Relación entre Ventas en Unidades y Ventas en Valor.png', dpi=300, bbox_inches='tight')
plt.show()

# Scatter Plot 2: Precio promedio vs. Ventas en unidades
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_consolidado, x='avg_price_per_unit', y='total_unit_sales', alpha=0.5)
plt.title('Relación entre Precio Promedio y Ventas en Unidades')
plt.xlabel('Precio Promedio por Unidad')
plt.ylabel('Ventas en Unidades')
plt.savefig('5.2-Relación entre Precio Promedio y Ventas en Unidades.png', dpi=300, bbox_inches='tight')
plt.show()

# Scatter Plot 3: Margen de beneficio vs. Ventas en valor
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_consolidado, x='profit_margin', y='total_value_sales', alpha=0.5)
plt.title('Relación entre Margen de Beneficio y Ventas en Valor')
plt.xlabel('Margen de Beneficio')
plt.ylabel('Ventas en Valor')
plt.savefig('5.3-Relación entre Margen de Beneficio y Ventas en Valor.png', dpi=300, bbox_inches='tight')
plt.show()

# Mapa de Calor de Correlaciones
plt.figure(figsize=(10, 8))
correlaciones = df_consolidado[['total_unit_sales', 'total_value_sales', 'avg_price_per_unit', 'profit_margin', 'total_unit_avg_weekly_sales']].corr()

sns.heatmap(correlaciones, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Mapa de Correlaciones entre Variables Numéricas')
plt.savefig('5.4-Mapa de Correlaciones entre Variables Numéricas.png', dpi=300, bbox_inches='tight')
plt.show()

# Filtro opcional para eliminar outliers extremos en scatter plot
df_filtrado = df_consolidado[(df_consolidado['total_unit_sales'] < 200) & (df_consolidado['total_value_sales'] < 2000)]

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_filtrado, x='total_unit_sales', y='total_value_sales', alpha=0.5)
plt.title('Relación Filtrada entre Ventas en Unidades y Ventas en Valor (<200 unidades y <2000 valor)')
plt.xlabel('Ventas en Unidades')
plt.ylabel('Ventas en Valor')
plt.savefig('5.5-Relación Filtrada entre Ventas en Unidades y Ventas en Valor.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Scatterplot de Ventas en Unidades vs. Ventas en Valor
Relación altamente lineal, lo que confirma que mayores unidades vendidas tienden a reflejarse en mayores valores de venta. Es normal ver esta relación si el precio promedio es relativamente constante.

### Scatterplot de Precio Promedio por Unidad vs. Ventas en Unidades
Relación negativa clara. A medida que el precio promedio sube, las unidades vendidas tienden a disminuir. Esto puede indicar sensibilidad al precio por parte de los consumidores.

### Scatterplot de Margen de Beneficio vs. Ventas en Valor
Relación altamente positiva y lineal, lo que sugiere que cuando hay más ventas en valor, el margen de beneficio tiende a ser mayor. Esto tiene sentido porque el margen puede estar basado en el valor total de las ventas.

### Mapa de Correlaciones
total_unit_sales y total_value_sales tienen correlación alta. 
avg_price_per_unit tiene correlación negativa débil con ventas en unidades.
profit_margin correlaciona bien con ventas en valor.
total_unit_avg_weekly_sales tiene correlaciones moderadas, lo cual es útil para análisis temporales.

### Scatterplot filtrado sin outliers extremos
Visualización mucho más limpia, sin los sesgos de los valores extremos. Permite ver de manera más clara la dispersión real de los datos en un rango más representativo.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Scatterplot con hue por Región
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df_consolidado, x='total_unit_sales', y='total_value_sales', hue='region', alpha=0.6)
plt.title('Relación entre Ventas en Unidades y Ventas en Valor por Región')
plt.xlabel('Ventas en Unidades')
plt.ylabel('Ventas en Valor')
plt.legend(title='Región', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.savefig('5.6-Relación entre Ventas en Unidades y Ventas en Valor por Región.png', dpi=300, bbox_inches='tight')
plt.show()

# Scatterplot con hue por Marca
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df_consolidado, x='total_unit_sales', y='total_value_sales', hue='brand', alpha=0.6)
plt.title('Relación entre Ventas en Unidades y Ventas en Valor por Marca')
plt.xlabel('Ventas en Unidades')
plt.ylabel('Ventas en Valor')
plt.legend(title='Marca', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.savefig('5.7-Relación entre Ventas en Unidades y Ventas en Valor por Marca.png', dpi=300, bbox_inches='tight')
plt.show()

# Regresión Lineal Simple
sns.lmplot(data=df_consolidado, x='total_unit_sales', y='total_value_sales', height=6, aspect=1.5, scatter_kws={'alpha': 0.3})
plt.title('Tendencia Lineal entre Ventas en Unidades y Ventas en Valor')
plt.xlabel('Ventas en Unidades')
plt.ylabel('Ventas en Valor')
plt.savefig('5.8-Tendencia Lineal entre Ventas en Unidades y Ventas en Valor.png', dpi=300, bbox_inches='tight')
plt.show()

# Regresión Lineal por Región
sns.lmplot(data=df_consolidado, x='total_unit_sales', y='total_value_sales', hue='region', height=6, aspect=1.5, scatter_kws={'alpha': 0.3})
plt.title('Tendencia Lineal entre Ventas en Unidades y Ventas en Valor por Región')
plt.xlabel('Ventas en Unidades')
plt.ylabel('Ventas en Valor')
plt.savefig('5.9-Tendencia Lineal entre Ventas en Unidades y Ventas en Valor por Región.png', dpi=300, bbox_inches='tight')
plt.show()

# Scatterplot y regresión lineal en escala logarítmica
plt.figure(figsize=(12, 8))
sns.regplot(
    x=np.log1p(df_consolidado['total_unit_sales']),
    y=np.log1p(df_consolidado['total_value_sales']),
    scatter_kws={'alpha': 0.3}
)
plt.title('Tendencia Lineal en Escala Logarítmica: Ventas en Unidades vs. Ventas en Valor')
plt.xlabel('Log Ventas en Unidades')
plt.ylabel('Log Ventas en Valor')
plt.savefig('5.10-Tendencia Lineal en Escala Logarítmica: Ventas en Unidades vs. Ventas en Valor.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Regresión lineal simple y por región

### Transformación logarítmica + regresión lineal

```python
# Punto 6: Visualizar la distribución geográfica de las ventas

import matplotlib.pyplot as plt
import seaborn as sns

# Agrupación por región y tiempo
ventas_region_tiempo = df_consolidado.groupby(['date', 'region']).agg({
    'total_value_sales': 'sum',
    'total_unit_sales': 'sum'
}).reset_index()

# Gráfico de Barras Apiladas
ventas_pivot = ventas_region_tiempo.pivot(index='date', columns='region', values='total_value_sales')
plt.figure(figsize=(16, 8))
ventas_pivot.plot(kind='bar', stacked=True, figsize=(16, 8), colormap='tab20')
plt.title('Distribución de Ventas en Valor por Región a lo largo del Tiempo')
plt.xlabel('Fecha')
plt.ylabel('Ventas en Valor')
plt.legend(title='Región', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=90, fontsize=8) # Ajustamos las fechas
plt.tight_layout() # Ajustamos las fechas
plt.savefig('6.1-Distribución de Ventas en Valor por Región a lo largo del Tiempo.png', dpi=300, bbox_inches='tight')
plt.show()

# Gráfico de Barras por Región
ventas_por_region = df_consolidado.groupby('region').agg({
    'total_value_sales': 'sum',
    'total_unit_sales': 'sum'
}).reset_index()
plt.figure(figsize=(12, 6))
sns.barplot(data=ventas_por_region, x='region', y='total_value_sales', palette='viridis')
plt.title('Ventas Totales en Valor por Región')
plt.xlabel('Región')
plt.ylabel('Ventas en Valor')
plt.xticks(rotation=45)
plt.savefig('6.2-Ventas Totales en Valor por Región.png', dpi=300, bbox_inches='tight')
plt.show()

# Pie Chart de Ventas
plt.figure(figsize=(8, 8))
plt.pie(ventas_por_region['total_value_sales'],
        labels=ventas_por_region['region'],
        autopct='%1.1f%%',
        startangle=140,
        colors=sns.color_palette('pastel'))
plt.title('Proporción de Ventas en Valor por Región')
plt.axis('equal')
plt.savefig('6.3-Proporción de Ventas en Valor por Región.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Gráfico de Barras Apiladas en el Tiempo
Las ventas de la región TOTAL AUTOS SCANNING MEXICO dominan claramente sobre el resto. La tendencia general es estable con algunos picos que podrían coincidir con temporadas específicas.

### Gráfico de Barras por Región
Se confirma que TOTAL AUTOS SCANNING MEXICO representa más del 50% de las ventas totales.

Muestra una distribución clara de la participación por región. TOTAL AUTOS SCANNING MEXICO representa el 50% de las ventas totales.
