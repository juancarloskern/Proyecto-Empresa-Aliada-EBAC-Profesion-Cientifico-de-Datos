```python
# Paso 1) Cargar las tablas de datos en Pandas
# Importación de bibliotecas
import pandas as pd
import numpy as np

# Ruta base
ruta_base = r"C:\Users\juanc\OneDrive\Documentos\2 Ciencia de Datos\EBAC México - Escuela Britanica de Artes Creativas y Tecnología\1 Profesión Cientifico de Datos\02 - Programación en Python para Ciencia de Datos\Proyecto empresa aliada entregable 1"

# Cargar archivos
try:
    df_category = pd.read_csv(rf"{ruta_base}\DIM_CATEGORY.csv") # Archivo CSV
    df_product = pd.read_excel(rf"{ruta_base}\DIM_PRODUCT.xlsx") # Archivo Excel
    df_segment = pd.read_excel(rf"{ruta_base}\DIM_SEGMENT.xlsx") # Archivo Excel
    df_calendar = pd.read_excel(rf"{ruta_base}\DIM_CALENDAR.xlsx") # Archivo Excel
    df_sales = pd.read_csv(rf"{ruta_base}\FACT_SALES.csv") # Archivo CSV
    print("Archivos cargados exitosamente.")
except Exception as e:
    print(f"Error al cargar los archivos: {e}")
```

```python
# Paso 2: Revisar y entender los datos cargados
# Inspeccionar las primeras filas y estructura de cada DataFrame
print('Calendario:')
print(df_calendar.head())
print(df_calendar.info())
print(df_calendar.describe())

print('Categoría:')
print(df_category.head())
print(df_category.info())
print(df_category.describe())

print('Producto:')
print(df_product.head())
print(df_product.info())
print(df_product.describe())

print('Segmento:')
print(df_segment.head())
print(df_segment.info())
print(df_segment.describe())

print('Ventas:')
print(df_sales.head())
print(df_sales.info())
print(df_sales.describe())
```

```python
# Checamos columnas
print("Columnas en df_sales:", df_sales.columns)
print("Columnas en df_product:", df_product.columns)
print("Columnas en df_category:", df_category.columns)
print("Columnas en df_segment:", df_segment.columns)
print("Columnas en df_calendar:", df_calendar.columns)
```

```python
# Paso 3: Limpieza de datos
# Limpiar nombres de columnas para evitar errores
df_sales.columns = df_sales.columns.str.strip().str.lower()
df_product.columns = df_product.columns.str.strip().str.lower()
df_category.columns = df_category.columns.str.strip().str.lower()
df_segment.columns = df_segment.columns.str.strip().str.lower()
df_calendar.columns = df_calendar.columns.str.strip().str.lower()

# Renombrar columnas para que coincidan en ambas tablas
df_sales.rename(columns={'item_code': 'item'}, inplace=True)
df_category.rename(columns={'id_category': 'category_id'}, inplace=True)

# Eliminar duplicados y manejar valores nulos
for df in [df_calendar, df_category, df_product, df_segment, df_sales]:
    # Eliminar duplicados
    df.drop_duplicates(inplace=True)
    # Llenar valores nulos hacia adelante
    df.ffill(inplace=True)

print("Datos limpios y sin duplicados.")
df
```

```python
# Checamos columnas
print("Columnas en df_sales:", df_sales.columns)
print("Columnas en df_product:", df_product.columns)
print("Columnas en df_category:", df_category.columns)
print("Columnas en df_segment:", df_segment.columns)
print("Columnas en df_calendar:", df_calendar.columns)
```

```python
# Paso 4: Unir los DataFrames relevantes
# Asegurar compatibilidad de tipos de datos antes de las uniones
df_sales['item'] = df_sales['item'].astype(str)
df_product['item'] = df_product['item'].astype(str)
df_category['category_id'] = df_category['category_id'].astype(str)
df_product['category'] = df_product['category'].astype(str)
df_segment['category'] = df_segment['category'].astype(str)
df_sales['region'] = df_sales['region'].astype(str)
df_calendar['week'] = df_calendar['week'].astype(str)

# Unir tablas usando claves comunes
merged_df = df_sales.merge(df_product, on='item', how='left')  # Unir productos
merged_df = merged_df.merge(df_category, left_on='category', right_on='category_id', how='left')  # Unir categorías
merged_df = merged_df.merge(df_segment, left_on='region', right_on='segment', how='left')  # Unir segmentos
merged_df = merged_df.merge(df_calendar, left_on='week', right_on='week', how='left')  # Unir calendario

# Mostrar información del DataFrame consolidado
print("Datos consolidados en un solo DataFrame:")
print(merged_df.info())
```

```python
# Paso 5: Transformaciones necesarias

# 5.1 Eliminar columnas duplicadas (_x y _y) que no son necesarias
merged_df.drop(columns=['category_x', 'format_x', 'attr1_x', 'attr2_x', 'attr3_x'], inplace=True)

# 5.2 Renombrar columnas para mayor claridad
merged_df.rename(columns={
    'category_y': 'category',  # Categoría consolidada
    'format_y': 'format',      # Formato consolidado
    'attr1_y': 'attr1',
    'attr2_y': 'attr2',
    'attr3_y': 'attr3'
}, inplace=True)

# 5.3 Crear columnas calculadas
# Calcular el precio promedio por unidad vendida
merged_df['avg_price_per_unit'] = merged_df['total_value_sales'] / merged_df['total_unit_sales']

# Crear columna de margen de ventas
merged_df['profit_margin'] = merged_df['total_value_sales'] * 0.2  # Si consideramos un 20% de margen de ganancia

# Crear columna de año-mes combinada para análisis temporal
merged_df['year_month'] = merged_df['year'].astype(str) + '-' + merged_df['month'].astype(str).str.zfill(2)

# 5.4 Estandarizar formatos
# Convertir fechas a formato datetime estándar (ya debería estar en datetime64, pero confirmamos)
merged_df['date'] = pd.to_datetime(merged_df['date'], errors='coerce')

# 5.5 Manejar columnas vacías (NaN)
# Eliminar columnas que están completamente vacías
merged_df.drop(columns=['category', 'attr1', 'attr2', 'attr3', 'format', 'segment'], inplace=True)

# 5.6 Manejar valores infinitos en columnas calculadas
# Reemplazar infinitos por NaN y luego llenarlos con 0
merged_df['avg_price_per_unit'] = merged_df['avg_price_per_unit'].replace([np.inf, -np.inf], np.nan)
merged_df['avg_price_per_unit'] = merged_df['avg_price_per_unit'].fillna(0)

# 5.7 Verificar valores atípicos (opcional)
print(merged_df[['total_unit_sales', 'total_value_sales']].describe(percentiles=[0.01, 0.99]))

# 5.8 Verificar los cambios realizados
print("Transformaciones realizadas. Vista previa del DataFrame:")
print(merged_df.head())
print(merged_df.info())
print(merged_df.describe())
```

```python
# Paso 6: Guardar el conjunto de datos consolidado
import os
output_path = rf"{ruta_base}\Consolidated_Data.csv"
merged_df.to_csv(output_path, index=False)

print(f"Consolidación completa. Archivo guardado en: {output_path}")
```
